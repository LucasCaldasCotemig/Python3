```python
path = r"C:\Users\LC\Downloads\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar

Exemplo 1

```python
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

Curso superior por região, ordenado pelo percentual de mulheres

```python
formacao_total.sort_values("percentual_mulheres", ascending=False)
```

Exemplo 2 - Ciência, Tecnologia, Engenharias e Matemática (CTEM)

```python
formacao_ctem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
```

```python
formacao_ctem.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_ctem
```

```python
formacao_ctem["percentual_mulheres"] = ((formacao_ctem['total_mulheres'] / formacao_ctem['total']) * 100).round(2)
```

```python
formacao_ctem.sort_values("percentual_mulheres", ascending=False)
```

Exemplo 3 - Educação, Serviços pessoais, Saúde e Bem-estar

```python
formacao_saude = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
```

```python
formacao_saude.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_saude
```

```python
formacao_saude["percentual_mulheres"] = ((formacao_saude['total_mulheres'] / formacao_saude['total']) * 100).round(2)
```

```python
formacao_saude.sort_values("percentual_mulheres", ascending=False)
```

Comparando a participação feminina nas três áreas

```python
comparacao = formacao_total[["uf_regiao", "percentual_mulheres"]].rename(
    columns={"percentual_mulheres": "pct_mulheres_superior"}
)
comparacao = comparacao.merge(
    formacao_ctem[["uf_regiao", "percentual_mulheres"]].rename(
        columns={"percentual_mulheres": "pct_mulheres_ctem"}
    ),
    on="uf_regiao",
)
comparacao = comparacao.merge(
    formacao_saude[["uf_regiao", "percentual_mulheres"]].rename(
        columns={"percentual_mulheres": "pct_mulheres_educ_saude"}
    ),
    on="uf_regiao",
)

brasil = comparacao.loc[comparacao["uf_regiao"] == "Brasil"].iloc[0]
print("Brasil - participação feminina por área de formação")
print(f"  Curso superior (geral):           {brasil.pct_mulheres_superior:.1f}%")
print(f"  Ciência, Tecnologia, Eng. e Mat.: {brasil.pct_mulheres_ctem:.1f}%")
print(f"  Educação, Serviços, Saúde e B-E.: {brasil.pct_mulheres_educ_saude:.1f}%")

comparacao.sort_values("pct_mulheres_ctem")
```
